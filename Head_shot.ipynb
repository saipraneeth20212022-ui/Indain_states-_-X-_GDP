```python
year_cols = ['2017-18', '2018-19', '2019-20', '2020-21', '2021-22', '2022-23', '2023-24']

# Convert to float first
for col in year_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').astype(float)

# Count NAs per row
df['na_count'] = df[year_cols].isnull().sum(axis=1)
print(df[['State/UnionTerritory', 'na_count']])
```

```python
df['na_count'] = df[year_cols].isnull().sum(axis=1)
print(df[['State/UnionTerritory', 'na_count']])
```

```python
# Case 1 — single NA rows → interpolate (uses both neighbours)
for idx in df.index:
    if df.loc[idx, 'na_count'] == 1:
        df.loc[idx, year_cols] = df.loc[idx, year_cols]\
            .interpolate(method='linear')

# Case 2 — multiple NA rows → growth rate from first 2 known years
for idx in df.index:
    if df.loc[idx, 'na_count'] > 1:
        # Calculate growth rate from 2017-18 to 2018-19
        base  = df.loc[idx, '2017-18']
        next_ = df.loc[idx, '2018-19']
        growth_rate = (next_ - base) / base  # e.g. 0.147 = 14.7%
        
        # Fill NAs forward using that growth rate
        for i in range(1, len(year_cols)):
            prev = year_cols[i-1]
            curr = year_cols[i]
            if pd.isna(df.loc[idx, curr]) and pd.notna(df.loc[idx, prev]):
                df.loc[idx, curr] = df.loc[idx, prev] * (1 + growth_rate)

# Verify
print(df[year_cols].isnull().sum())
print(df[['State/UnionTerritory'] + year_cols])
```

```python
df = df.drop(columns=df.columns[-1])
print(df.columns.tolist())  # verify
```

```python
year_cols = df.columns[1:].tolist()

# Step 1 - calculate % growth year by year
for i in range(1, len(year_cols)):
    prev = year_cols[i-1]
    curr = year_cols[i]
    df[f'pct_growth_{curr}'] = ((df[curr] - df[prev]) / df[prev]) * 100

growth_cols = [col for col in df.columns if col.startswith('pct_growth_')]

# Step 2 - avg growth per state
df['Avg_Growth'] = df[growth_cols].mean(axis=1)

# Step 3 - report
report = df[['State/UnionTerritory', 'Avg_Growth']]\
         .sort_values('Avg_Growth', ascending=False)\
         .reset_index(drop=True)
print(report)
```

```python
# Step 4 - plot
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 8))

colors = ['green' if x > report['Avg_Growth'].mean() 
          else 'steelblue' for x in report['Avg_Growth']]

ax.barh(report['State/UnionTerritory'], report['Avg_Growth'], color=colors)
ax.axvline(x=report['Avg_Growth'].mean(), color='red',
           linestyle='--', label=f"Overall Mean: {report['Avg_Growth'].mean():.2f}%")

for i, v in enumerate(report['Avg_Growth']):
    ax.text(v + 0.1, i, f'{v:.2f}%', va='center', fontsize=8)

ax.set_xlabel('Average Growth %')
ax.set_title('Average GDP Growth Rate of Indian States (2017-18 to 2023-24)')
ax.legend()
ax.invert_yaxis()
plt.tight_layout()
plt.savefig('gdp_growth_rate.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Standard deviation tells consistency
# Low std = consistent, High std = ups and downs

df['Std_Growth'] = df[growth_cols].std(axis=1)

overall_avg = df['Avg_Growth'].mean()
overall_std = df['Std_Growth'].mean()

print(f"Overall Avg Growth : {overall_avg:.2f}%")
print(f"Overall Avg Std    : {overall_std:.2f}%")
```

```python
def classify(row):
    if row['Avg_Growth'] >= overall_avg and row['Std_Growth'] <= overall_std:
        return 'Consistently Fast'          # High avg + Low std
    elif row['Avg_Growth'] < overall_avg and row['Std_Growth'] <= overall_std:
        return 'Consistently Struggling'    # Low avg  + Low std
    elif row['Avg_Growth'] >= overall_avg:
        return 'Fast but Inconsistent'      # High avg + High std
    else:
        return 'Struggling and Unstable'    # Low avg  + High std ← fixed name!
df['Category'] = df.apply(classify, axis=1)
print(df[['State/UnionTerritory', 'Avg_Growth', 'Std_Growth', 'Category']])
```

```python
# Print each category
for cat in ['Consistently Fast', 'Consistently Struggling', 
            'Fast but Inconsistent', 'Struggling and Unstable']:  # ← fixed name
    states = df[df['Category'] == cat]['State/UnionTerritory'].tolist()
    print(f"\n{cat}:")
    print(states)

# Plot
import matplotlib.pyplot as plt

colors = {
    'Consistently Fast'      : 'green',
    'Consistently Struggling': 'red',
    'Fast but Inconsistent'  : 'orange',
    'Struggling and Unstable': 'steelblue'   # ← fixed name
}

sorted_df = df.sort_values('Avg_Growth', ascending=False)

fig, ax = plt.subplots(figsize=(12, 8))

for category, color in colors.items():
    subset = sorted_df[sorted_df['Category'] == category]
    ax.barh(subset['State/UnionTerritory'], subset['Avg_Growth'],
            color=color, label=category)

# mean line
ax.axvline(x=overall_avg, color='black', linestyle='--',
           label=f'Overall Mean: {overall_avg:.2f}%')

# value labels
for i, (_, row) in enumerate(sorted_df.iterrows()):
    ax.text(row['Avg_Growth'] + 0.1, i,
            f"{row['Avg_Growth']:.2f}%", va='center', fontsize=8)

ax.set_xlabel('Average Growth %')
ax.set_title('Consistently Fast vs Struggling Indian States')
ax.legend(loc='lower right')
ax.invert_yaxis()

plt.tight_layout()
plt.savefig('gdp_growth_story.jpg', dpi=300, bbox_inches='tight')
plt.show()
```
